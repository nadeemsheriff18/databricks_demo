```python
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")

spark.sql(f"""

CREATE OR REPLACE TABLE
{catalog}.{schema}.sales_job_demo

AS

SELECT *

FROM read_files(
'/Volumes/nadeem_catalog/ingestion_demo/raw_files/orders.json',
format=>'csv',
header=>true,
inferSchema=>true
)

""")
```

```python
count = spark.table(
    "nadeem_catalog.ingestion_demo.sales_job_demo"
).count()
print("Module 4 Demoo")
dbutils.jobs.taskValues.set(
    key="record_count",
    value=count
)
```
